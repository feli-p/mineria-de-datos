Principales resultados por localidad (ITER) del Censo de Población y Vivienda 2020.

Datos: <https://www.inegi.org.mx/programas/ccpv/2020/#datos_abiertos>

```python
import pandas as pd

ruta = "datos/datos_crudos/iter_00_cpv2020/conjunto_de_datos/conjunto_de_datos_iter_00CSV20.csv"

iter_df = pd.read_csv(ruta, sep=",", encoding="utf-8", dtype=str)
print("Número de columnas:", len(iter_df.columns))
print(iter_df.columns.tolist())

columnas_relevantes = [c for c in iter_df.columns if 'P_0A4' in c.upper()]
print(columnas_relevantes)
```

```python
municipios_iter = iter_df[
    (iter_df['LOC'] == '0000') &   # solo totales, no localidades individuales
    (iter_df['MUN'] != '000')      # excluye los totales estatales/nacional
].copy()

print(municipios_iter.shape)
```

## Coneval

    "Estimaciones de área pequeña de CONEVAL, calibradas para que la suma de población municipal coincida con la población estatal reportada en el MCS-ENIGH/Modelo Estadístico correspondiente. No es un conteo directo — la población en este archivo puede diferir de la reportada por INEGI o CONAPO a nivel municipal."

datos: <https://www.coneval.org.mx/Medicion/Paginas/Programas_BD_municipal_2010_2020.aspx>

```python
import openpyxl

archivo = "datos/datos_crudos/coneval/Concentrado_indicadores_de_pobreza_2020.xlsx"

wb = openpyxl.load_workbook(archivo, read_only=True)
print(wb.sheetnames)
```

```python
import pandas as pd

archivo = "datos/datos_crudos/coneval/Concentrado_indicadores_de_pobreza_2020.xlsx"
hoja = "Concentrado municipal"

crudo = pd.read_excel(archivo, sheet_name=hoja, header=None)

fila_id_grupo = crudo.iloc[4].ffill()   # rellena "Pobreza", "Pobreza extrema"... hacia la derecha
fila_subdivision = crudo.iloc[5]

nombres_finales = []
columnas_id = ['Clave de entidad', 'Entidad federativa', 'Clave de municipio', 'Municipio']

for i in range(len(crudo.columns)):
    valor_id = crudo.iloc[4, i]
    if pd.notna(valor_id) and str(valor_id).strip() in columnas_id:
        # columnas de identificación: usan directamente su nombre
        nombres_finales.append(str(valor_id).strip())
    else:
        grupo = str(fila_id_grupo[i]).strip() if pd.notna(fila_id_grupo[i]) else ""
        sub = str(fila_subdivision[i]).replace("\n", " ").strip() if pd.notna(fila_subdivision[i]) else ""
        nombres_finales.append(f"{grupo} - {sub}".strip(" -"))

print(nombres_finales[:15])   # revisa que las primeras 15 se vean bien
print(len(nombres_finales))   # debería ser 146, como viste en el shape
```

```python
coneval = crudo.iloc[8:2477].copy()
coneval.columns = nombres_finales
coneval = coneval.drop(columns=[''])
coneval = coneval.reset_index(drop=True)

print(coneval.shape)          # esperamos (2469, 146)
print(coneval.head())
```

```python
import pandas as pd
import glob

archivos_comercio = sorted(glob.glob("datos/datos_crudos/denue_temp/*_46*/conjunto_de_datos/*.csv"))
print("Archivos encontrados:", archivos_comercio)

codigos_buscados = ['464111', '464112', '462111']

for archivo in archivos_comercio:
    df = pd.read_csv(archivo, encoding="latin-1", dtype=str, usecols=['codigo_act'])
    encontrados = df[df['codigo_act'].isin(codigos_buscados)]
    print(archivo, "->", encontrados['codigo_act'].value_counts().to_dict())
```

```python
import pandas as pd

columnas_necesarias = ['id', 'codigo_act', 'nombre_act', 'cve_ent', 'cve_mun', 'cve_loc']

archivos_relevantes = {
    "supermercados": "datos/datos_crudos/denue_temp/denue_00_46112-46311_1120_csv/conjunto_de_datos/denue_inegi_46112-46311_.csv",
    "farmacias": "datos/datos_crudos/denue_temp/denue_00_46321-46531_1120_csv/conjunto_de_datos/denue_inegi_46321-46531_.csv",
    "guarderias": "datos/datos_crudos/denue_temp/denue_00_62_1120_csv/conjunto_de_datos/denue_inegi_62_.csv",
}

codigos_finales = ['462111', '464111', '464112', '624411', '624412']

dfs = []
for nombre, ruta in archivos_relevantes.items():
    df = pd.read_csv(ruta, encoding="latin-1", dtype=str, usecols=columnas_necesarias)
    df_filtrado = df[df['codigo_act'].isin(codigos_finales)]
    dfs.append(df_filtrado)
    print(nombre, "->", df_filtrado.shape)

denue_final = pd.concat(dfs, ignore_index=True)
print("Total consolidado:", denue_final.shape)
print(denue_final['codigo_act'].value_counts())
```
